```python
import torch
import spikingjelly.clock_driven.neuron as neuron
import spikingjelly.clock_driven.encoding as encoding
import spikingjelly.clock_driven.functional as functional
import spikingjelly.clock_driven.surrogate as surrogate

# Create a simple LIF neuron
lif_neuron = neuron.LIFNode(surrogate_function=surrogate.ATan())

# Create a simple input encoder (Poisson encoder)
encoder = encoding.PoissonEncoder()

# Generate a sample input (random data)
input_data = torch.rand(100)

# Encode the input data into spike trains
encoded_data = encoder(input_data)

# Simulate the neuron with the encoded spike trains
for t in range(100):
    lif_neuron(encoded_data[t])
    print(f"Time step {t}, Membrane potential: {lif_neuron.v.item()}")

    # Reset the neuron state after each time step
    functional.reset_net(lif_neuron)
```

```python
import torch
import torch.nn as nn
import spikingjelly.clock_driven.neuron as neuron
import spikingjelly.clock_driven.encoding as encoding
import spikingjelly.clock_driven.functional as functional
import spikingjelly.clock_driven.layer as layer
import spikingjelly.clock_driven.surrogate as surrogate
```

```python
class SimpleSNN(nn.Module):
    def __init__(self):
        super(SimpleSNN, self).__init__()
        self.encoder = encoding.PoissonEncoder()
        self.fc1 = nn.Linear(28 * 28, 100)  # Input layer to hidden layer
        self.lif1 = neuron.LIFNode(surrogate_function=surrogate.ATan())
        self.fc2 = nn.Linear(100, 10)  # Hidden layer to output layer
        self.lif2 = neuron.LIFNode(surrogate_function=surrogate.ATan())

    def forward(self, x):
        x = self.encoder(x).float()  # Encode the input
        x = x.view(x.size(0), -1)  # Flatten the input
        x = self.fc1(x)
        x = self.lif1(x)
        x = self.fc2(x)
        x = self.lif2(x)
        return x

# Instantiate the network
net = SimpleSNN()
```

```python
input_data = torch.rand((1, 28, 28))  # Example input: a single 28x28 image
```

```python
# Number of time steps to simulate
num_steps = 100

# Simulate the network
for step in range(num_steps):
    output = net(input_data)
    print(f"Time step {step + 1}, Output: {output}")

    # Reset the network state after each step
    functional.reset_net(net)
```
